```python
import cv2
import os

# ==============================
# 📁 PATHS
# ==============================
video_path = r"D:\IPCV\LAB-EXP-15\video\input.mp4"
images_folder = r"D:\IPCV\LAB-EXP-15\images"
result_folder = r"D:\IPCV\LAB-EXP-15\resultant_img"

os.makedirs(images_folder, exist_ok=True)
os.makedirs(result_folder, exist_ok=True)

# ==============================
# 🎥 LOAD VIDEO
# ==============================
cap = cv2.VideoCapture(video_path)

# Background subtractor (motion detection)
fgbg = cv2.createBackgroundSubtractorMOG2()

frame_count = 0

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Resize (optional for speed)
    frame = cv2.resize(frame, (500, 400))

    # Convert to grayscale
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Apply background subtraction
    fgmask = fgbg.apply(gray)

    # Threshold to remove noise
    _, thresh = cv2.threshold(fgmask, 200, 255, cv2.THRESH_BINARY)

    # Find contours (moving objects)
    contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    # Draw bounding boxes
    for cnt in contours:
        if cv2.contourArea(cnt) > 500:   # filter small noise
            x, y, w, h = cv2.boundingRect(cnt)
            cv2.rectangle(frame, (x,y), (x+w, y+h), (0,255,0), 2)

    # ==============================
    # 💾 SAVE INPUT & OUTPUT FRAMES
    # ==============================
    if frame_count < 5:  # save first 5 frames
        cv2.imwrite(os.path.join(images_folder, f"input_{frame_count}.png"), frame)
        cv2.imwrite(os.path.join(result_folder, f"output_{frame_count}.png"), frame)

    frame_count += 1

cap.release()

print("✅ Frames saved successfully!")
```
